**`tbb` type aliases**

```c++
using Arena = tbb::task_arena;
using Range = tbb::blocked_range<size_t>;
```

**parallel initialization**

```c++
%main
auto values = std::vector<double>(1e7);
Arena(4).execute([&values](){
tbb::parallel_for(Range(0,values.size(),2.5e6),
  [&values](const Range& r){
    auto idx = tbb::this_task_arena::current_thread_index();
    std::println("t: {:2d}, from {:8d} to {:8d} ",idx, r.begin(), r.end());
    for (auto i=r.begin(); i<r.end(); ++i)
      values[i] = 1.0;
  });
});
```
